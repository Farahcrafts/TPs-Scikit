```python
# meme que a moi je veux pas supprimer cette ligne car il y a juste 2 valeurs manquantes
data=data.dropna()
data.info()
```

```python
# je propose qu lieu de retirer les valeurs negatives pour fare_amount, on les corrige avec abs()
data["fare_amount"]= abs(data["fare_amount"])
print(data["fare_amount"].mean())
```

```python
#fixer les datetimes
data["pickup_datetime"]= pd.to_datetime(data["pickup_datetime"], format = "mixed" , errors="coerce")
print(type(data["pickup_datetime"]))
```
